```python
-- Executive Portfolio Overview Metrics

SELECT 
    COUNT(ID) AS total_customers,
    ROUND(AVG(LIMIT_BAL), 2) AS avg_credit_limit,
    SUM(LIMIT_BAL) AS total_credit_exposure,
    ROUND(AVG(default_payment_next_month) * 100, 2) AS default_rate_pct,
    COUNTIF(PAY_1 > 0) AS watchlist_customers,
    ROUND(COUNTIF(PAY_1 > 0) * 100.0 / COUNT(ID), 2) AS watchlist_pct
FROM `jiwambe_analytics.credit_card_clients`;
```
